# Sanity check przychodnia - wariant z umówieniami

## Dokumentacja kodu

```python
import matplotlib
matplotlib.use('TkAgg')  # or 'Qt5Agg'
import numpy as np
import simpy
import matplotlib.pyplot as plt
import random
import pandas as pd
from pomocnicze_funkcje import readable_time, randomized_service_time
```

## Parametry klasy Patient:
- **id**: identyfikator pacjenta
- **arrival_time**: czas przybycia pacjenta do kliniki
- **service_start_time**: czas rozpoczęcia wizyty
- **service_end_time**: czas zakonczenia wizyty
- **room**: identyfikator gabinetu przypisanego pacjentowi
- **real_service_time**: przechowuje liczbe pseudolosową wygenerowaną z rozkładu normalnego odpowiadającą długości wizyty

```python
class Patient:
    def __init__(self, id):
        self.id = id
        self.arrival_time = None
        self.service_start_time = None
        self.service_end_time = None
        self.room = None
        self.real_service_time = None
```

## Parametry klasy Gabinet:
- **id**: identyfikator gabinetu
- **env**: środowiko symulacji
- **patients_served**: ilość obsłużonych pacjentów
- **queue**: kolejka do gabinetu
- **queue_len_dict**: pozwala zapisywać długość kolejki w różnych momentach sumulacji
- **resource**: wskazuje na obiekt z biblioteki sympy który odpowiada za faktyczne obsługiwanie symulacji

Oraz funkcja **update_queue_len_dict** która odpowiada za zapisywanie długości kolejki w danym momencie czasu do słownika **queue_len_dict**

```python
class Room:
    def __init__(self, id, env):
        self.id = id
        self.patients_served = 0
        self.no_show = 0
        self.queue = []
        self.queue_len_dict = dict()
        self.resource = simpy.Resource(env, capacity=1)
    def update_queue_len_dict(self, curr_time):
        self.queue_len_dict[curr_time] = len(self.queue)
```

## Parametry klasy Clinic:
- **curr_patient_id**: aktualny identyfikator pacjenta u żywany przy tworzeniu pacjentów
- **env**: środowsko symuacji
- **service_time**: czas obsługi jednego pacjanta
- **no_show**: prawdopodobieństwo że pacjent nie zjawi na wizytę
- **seed**: seed generatora liczby pseudolosowych
- **list_rooms**: lista gabinetów
- **processed_patients**: lista, która przechowuje informacje o pacjkentach którzy już zostali obsłużeni
- **sim_time**: całkowity czas symulacji

```python
class Clinic:
    def __init__(self, number_of_rooms, planned_service_time, no_show_proportion=0.2, seed=None, sim_time=480):
        # zapisanie parametrów 
        self.env = simpy.Environment()
        self.curr_patient_id = 1
        self.planned_service_time = planned_service_time
        self.no_show_proportion = no_show_proportion
        self.sim_time = sim_time
        # utworzenie pokoi
        self.list_rooms = [Room(id=i + 1, env=self.env) for i in range(number_of_rooms)]
        self.processed_patients = []
        

        self.seed = seed
        if self.seed:
            random.seed(self.seed)
            np.random.seed(self.seed)
```

Metoda **generate_patients** jako wejście przyjmuje instancje klasy **room** i służy do generowania pacjantów i nadawania im atrybutów taklich jak czas obsługi (**real_service_time**) generowany z rozkładu normalnego oraz gabinet z którego będzie korzystał. Pacjeci są genrowani co stały czas **planned_service_time** i losuje czy osoba się pojawi czy wystąpi no show. 
Następnie przekazuje pacjenta do metody **serve_patient**. 

Jeżeli czas symulacji będzię wyższy niż aktualnego czasu + planowanego czasu obsługi to generowanie pacjętów zostaje przerwane aby nie doszło do sytułacji w której pacjęci są obsługiwani po zakończeniu pracy przychodni

```python
def generate_patients(self, room, verbose=False):

        def time_between_new_patients(): # funkcja podajaca czas miedzy wizytami
            return self.planned_service_time

        while True:
            if self.env.now > self.sim_time - self.planned_service_time:  # Pacjenci nie przychodzą przed {planned_service_time} zamknięciem
                break
            
            # tworzymy pacjenta i nadajemu mu atrybuty
            patient = Patient(id=f"{room.id}.{self.curr_patient_id}")
            patient.room = room.id
            patient.real_service_time = randomized_service_time()

            #sprawdzamy no_show
            if random.random() < self.no_show_proportion:  #pacjent nie przyszedl
                if verbose:
                    print(f'Czas {readable_time(self.env.now)}: Pacjent {patient.id} nie pojawił w gabinecie {room.id} w ciągu {self.planned_service_time} minut')
                room.no_show += 1


            else:  #pacjent przyszedl
                patient.arrival_time = self.env.now
                if verbose:
                    print(f"Czas {readable_time(self.env.now)}: Pacjent {patient.id} przybył do kliniki")
                self.env.process(self.serve_patient(patient, room, verbose))

            self.curr_patient_id += 1
            yield self.env.timeout(time_between_new_patients())
Clinic.generate_patients = generate_patients
```

Metoda **serve_patient** służy do operowania kolejką do gabinetu samą obsługą pacjenta oraz jego wyjściem z gabientu. Zapisuje również stany kolejki.

```python
def serve_patient(self, patient, room, verbose=False):
        # pacjent wchodzi do kolejki
        room.queue.append(patient.id)
        room.update_queue_len_dict(curr_time=self.env.now)

        with room.resource.request() as request:
            yield request
            # pacjent wychodzi z kolejki i wchodzi do gabinetu
            patient.service_start_time = self.env.now
            room.queue.remove(patient.id)
            room.update_queue_len_dict(curr_time=self.env.now)
            if verbose:
                print(f"Czas {readable_time(self.env.now)}: Pacjent {patient.id} wchodzi do gabinetu {room.id} ")
            yield self.env.timeout(patient.real_service_time)
            # pacjent wychodzi z gabinetu
            patient.service_end_time = self.env.now
            room.patients_served += 1
            if verbose:
                print(f"Czas {readable_time(self.env.now)}: Pacjent {patient.id} wychodzi z gabinetu {room.id}")
            self.processed_patients.append(patient)
Clinic.serve_patient = serve_patient
```

Metoda **reset** służy do resetowania kliniki między kolejnymi wywołaniami polecenia **run** w metoodzie **run_multiple_times**

```python
def reset(self):
    self.env = simpy.Environment()
    self.curr_patient_id = 1
    self.list_rooms = [Room(id=i + 1, env=self.env) for i in range(len(self.list_rooms))]
    self.processed_patients = []
Clinic.reset = reset
```

Metoda **run** odpowiada za uruchomienie symulacji dla jednego dnia po wywołaniu dla każdego gabinetu rozpoczyna osobny generator pacjentów.

Czas trwania jest wydłużony o 0.01 aby komunikaty które pojawiały się w ostatniej "minucie" symulacji mogły się wyświetlić.

Zawiera również funkcje **sum_queue_len_data** która sumuje stany kolejki ze wszystkich gabinetów.

I ostatecznie po zakonczeniu symulacji zapisuje zarówno dane o długości koleji oraz dane o obsłużonych pacjentach do DataFrame.

```python
def run(self, day=1, verbose=False):
        # funkcja do laczenia slownikow z kolejkami w jednen DataFrame
        def sum_queue_len_data():
            all_times = set()
            for room in self.list_rooms:
                all_times.update(room.queue_len_dict.keys())

            all_times = sorted(all_times)
            queue_data = []
            last_known = {room.id: 0 for room in self.list_rooms}

            for time in all_times:
                for room in self.list_rooms:
                    if time in room.queue_len_dict:
                        last_known[room.id] = room.queue_len_dict[time]

                total_queue = sum(last_known.values())
                queue_data.append({
                    'day': day,
                    'time': time,
                    'total_queue_length': total_queue
                })
            queue_data = pd.DataFrame(queue_data)
            queue_data = queue_data.set_index(["day", "time"])
            return queue_data
        
        # Dla kazdego pokoju uruchamiamy osobny proces generowania i obslugi pacjentow
        for room in self.list_rooms:
            self.env.process(self.generate_patients(room, verbose))
        # Symulacja trwa do ustalonego czasu + 0.01 minuity aby rekordy ostatnich pacjentow zdarzyły zostać zapisane
        self.env.run(until=self.sim_time + 0.01)
        # Zapisujemy do pandas DataFrame
        df = pd.DataFrame([
            {
                "day": day,
                "id": p.id,
                "room": p.room,
                "arrival_time": p.arrival_time,
                "service_start_time": p.service_start_time,
                "service_end_time": p.service_end_time,
                "service_time": p.service_end_time - p.service_start_time,
                "waiting_time": p.service_start_time - p.arrival_time
            }
            for p in self.processed_patients
        ])
        df = df.set_index("id")
        queue_df = sum_queue_len_data()
        return df, queue_df
Clinic.run = run
```

Metoda **run_multiple_times**

Przyjmuje n jako zmienną, która określa ilość wykonanych symulacji.

Łączy równierz dane z wszystkich wywołań **run** w jeden DataFrame.

Oraz dodatkowo zlicza ilość pacjentów ogółem, pacjentów obsłużonych, pacjentów no-show oraz proporcje no-show do wszystkich.

```python
def run_multiple_times(self, n, verbose=False, verbose_proportions=False):

        sum_no_show = 0
        sum_patients_served = 0

        for i in range(1, n + 1):
            if verbose:
                print(f"-----------{i}-----------")
            self.reset()
            single_df, single_queue_df = self.run(i, verbose)

            for room in self.list_rooms:
                sum_no_show += room.no_show
                sum_patients_served += room.patients_served

            multiple_df = pd.concat([multiple_df, single_df], ignore_index=True) if i > 1 else single_df
            multiple_queue_df = pd.concat([multiple_queue_df, single_queue_df]) if i > 1 else single_queue_df
        if verbose and verbose_proportions:
            print()
        if verbose_proportions:
            print(f"wszyscy pacjeci: {sum_patients_served + sum_no_show}")
            print(f"obsluzeni pacjeci: {sum_patients_served}")
            print(f"noshow pacjeci: {sum_no_show}")
            print(f"proporcja: {sum_no_show/(sum_patients_served + sum_no_show)}")
        return multiple_df, multiple_queue_df
Clinic.run_multiple_times = run_multiple_times
```
